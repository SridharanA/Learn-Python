**Example:**

Data in populations.txt describes the populations of hares(rabbits), lynxes(type of wild cats) and carrots in northern Canada during 20 years.

```python
# load txt data into numpy array object
data = np.loadtxt('populations.txt')
```

```python
data
```

```python
year, hares, lynxes, carrots = data.T # columns to variables
```

```python
year
```

```python
populations = data[:, 1:]   # neglecting the year and only taking the populations of hares, lynxes and rabbits
populations
```

```python
# simple std
populations.std(axis=0) # first axis - column-wise - std
```

```python
# Question: Which species has the highest population each year?
max_population_year_wise_index = populations.argmax(axis=1) # row-wise - maximum value
max_population_year_wise_index
```

```python
print('Year', 'Category', 'Population')

category = {
            0: 'Hares', 
            1: 'Lynxes',
            2: 'Carrots'
           }

for i in range(20):
    print(int(data[i, 0]), category[max_population_year_wise_index[i]], int(data[i, max_population_year_wise_index[i]]))
```

**Reshaping**

The inverse operation to flattening

```python
a = np.array([[1, 2, 3], [4, 5, 6]])

print(a)
```

```python
b = a.ravel()

b
```

```python
b = b.reshape((2, 3))

b
```

```python
b[0, 0] = 100

b
```

**Note and       Beware: reshape may also return a copy!:**

```python
a = np.zeros((3, 2))

print(a) # 3x2

b = a.T

print(b) # 2x3

b = b.reshape((3, 2))

print(b) # 3x2
```

**Adding a Dimension**

Indexing with the np.newaxis object allows us to add an axis to an array

newaxis is used to increase the dimension of the existing array by one more dimension, when used once. Thus,

1D array will become 2D array

2D array will become 3D array

3D array will become 4D array and so on

```python
z = np.array([1, 2, 3, 4])
z
```

```python
z = z[:, np.newaxis]

print(z)
print(z.ndim)
print(z.shape)
```

**Dimension Shuffling**

```python
a = np.arange(1, 3*4*3+1).reshape((3, 4, 3))

print(a)
print(a.ndim)
print(a.shape)
```

```python
a[0, 3, 2]
```

**Resizing**

```python
a = np.arange(4)

print(a)
print(a.shape)

a.resize((7,))

print(a)
print(a.shape)
```
