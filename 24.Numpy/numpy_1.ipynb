# 3. Indexing & Slicing

**3.1 Indexing**

The items of an array can be accessed and assigned to the same way as other **Python sequences (e.g. lists)**:

```python
a = np.arange(10)

print(a[5])  #indices begin at 0, like other Python sequences (and C/C++)
```

```python
# For multidimensional arrays, indexes are tuples or lists of integers

a = np.diag([1, 2, 3])

print(a)

print(a[1, 1])

print(a[1]) # returning row with index 1 in the numpy array
```

```python
a[(2, 1)] = 5 # assigning value

a
```

```python
a[[0,1]] = 50 # assigning value

a
```

```python
a[(0,1)] = 115 # assigning value

a
```

**3.2 Slicing**

```python
a = np.arange(10)

a
```

```python
a[1:7:2] # slicing - array_name[start_index:end_index(exclusive):step_size]
```

```python
# we can also combine assignment and slicing

a[5:] = 10

a
```

```python
b = np.arange(21, 26)
```

```python
a[5:] = b[::-1] # reverse assignment of numpy array 'b' to numpy array 'a'

a
```

# 4. Copies and Views

A slicing operation creates a view on the original array, which is just a way of accessing array data. Thus the original array is not copied in memory. You can use **np.shares_memory()** to check if two arrays share the same memory block. 

**When modifying the view, the original array is modified as well:**

```python
a = np.arange(10)

a
```

**shallow copy**

```python
b = a[::2]

b
```

```python
np.shares_memory(a, b)
```

```python
id(a[0]), id(b[0])
```

```python
id(a[4]), id(b[4])
```

```python
b[0] = 10

print(b)
```

```python
a # eventhough we modified b,  it updated 'a' because both shares same memory
```

**deep copy**

```python
a = np.arange(10)

a
```

```python
c = a[::2].copy() # force a copy

c
```

```python
np.shares_memory(a, c)
```

```python
id(a[0]), id(c[0]) # below no. thousand every number is only created once and referenced but changes in one will not get reflected in another
```

```python
c[0] = 90
```

```python
c
```

# 5. Fancy Indexing

NumPy arrays can be indexed with slices, but also with boolean or integer arrays **(masks)**. This method is called **fancy indexing**. **It creates copies not views**.

**Using Boolean Mask**

```python
a = np.random.randint(1, 20, 15)

a
```

```python
mask = (a % 2 == 0)
mask
```

```python
lst = [] # emtpy list

for x in mask:
    if x:
        lst.append(1)
    else:
        lst.append(0)

print(lst)

mask01 = np.array(lst)
mask01
```

```python
result1 = a[mask] # mask with boolean values (True or False) returns the values in that numpy array where they are True only
result1
```

```python
result2 = a[mask01] # mask with 0 or 1 returns the values in numpy array with index 0 or 1
result2
```

```python
extract_from_a = a[mask]

extract_from_a
```

**Indexing with a mask can be very useful to assign a new value to a sub-array:**

```python
a[mask] = -1 # mask is marked True where elements in numpy array are even number so they will be assigned -1

a
```
